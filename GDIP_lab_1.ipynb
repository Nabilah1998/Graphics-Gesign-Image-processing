```python
from google.colab import files
uploaded = files.upload()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread(list(uploaded.keys())[0])
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
h, w = image.shape[:2]
tx, ty = 100, 50   # Translation values
M = np.float32([[1, 0, tx], [0, 1, ty]])
translated = cv2.warpAffine(image, M, (w, h))
```

```python
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(translated)
plt.title("Translated Image")
plt.axis("off")

plt.show()
```

```python
# Combined Transformation
# Step 1: Upload Image
from google.colab import files
uploaded = files.upload()

# Step 2: Import Libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Step 3: Read Uploaded Image
image = cv2.imread(list(uploaded.keys())[0])
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

h, w = image.shape[:2]

# -------------------------------
# Step 4: Scaling
sx, sy = 1.2, 1.2   # image big
scaled = cv2.resize(image, None, fx=sx, fy=sy)

# Step 5: Rotation
h2, w2 = scaled.shape[:2]
center = (w2 // 2, h2 // 2)
M_rotation = cv2.getRotationMatrix2D(center, 45, 1.0)   # 45 degree rotation
rotated = cv2.warpAffine(scaled, M_rotation, (w2, h2))

# Step 6: Translation
tx, ty = 50, 30
M_translation = np.float32([[1, 0, tx], [0, 1, ty]])
transformed = cv2.warpAffine(rotated, M_translation, (w2 + tx, h2 + ty))

# -------------------------------
# Step 7: Display
plt.figure(figsize=(15, 5))

plt.subplot(1, 7, 1)
plt.imshow(image)
plt.title("Original")
plt.axis("off")

plt.subplot(1, 7, 2)
plt.imshow(translated)
plt.title("Translated Image")
plt.axis("off")

plt.subplot(1, 7, 3)
plt.imshow(scaled)
plt.title("Scaled Image")
plt.axis("off")

plt.subplot(1, 7, 4)
plt.imshow(rotated)
plt.title("After Rotation")
plt.axis("off")

plt.subplot(1, 7, 5)
plt.imshow(reflected_x)
plt.title("Reflection X-axis")
plt.axis("off")

plt.subplot(1, 7, 6)
plt.imshow(reflected_y)
plt.title("Reflection Y-axis")
plt.axis("off")

plt.subplot(1, 7, 7)
plt.imshow(transformed)
plt.title("Final Transformation")
plt.axis("off")

plt.show()
```
